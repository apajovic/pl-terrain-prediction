```python

from skimage.io import imread
import matplotlib.pyplot as plt
```

```python
BLACK_PL_PATH = './data/base/out/no_obstacle.png'
IMAGE_PATH = './output/unet_static/predictions/model_pred_wrap002.png'
IMAGE2_PATH = './output/pmnet_v3/model_pred_wrap002.png'

CONTROL_PATH = './output/unet_static/targets/model_pred_wrap002.png'
```

```python

def add_L(image_path,image2_path, control_path, black_pl_path):
    image = imread(image_path)
    image2 = imread(image2_path)
    control = imread(control_path)
    black_pl = imread(black_pl_path)
    control_pl = black_pl + control
    image_pl = black_pl + image
    image2_pl = black_pl + image2
    return image_pl, image2_pl, control_pl
```

```python
def show_comparisons(im_num=''):
    impath = f'./output/unet_static/predictions/model_pred_wrap{im_num}.png'
    im2path = f'./output/pmnet_v3/model_pred_wrap{im_num}.png'
    controlpath = f'./output/unet_static/targets/model_pred_wrap{im_num}.png'
    
    fig, axs = plt.subplots(2, 3, figsize=(10, 10))

    image_pl, image2_pl, control_pl = add_L(impath, im2path, controlpath, BLACK_PL_PATH)
    image, image2, control = imread(impath), imread(im2path), imread(controlpath)

    fig.suptitle(f'wrap{im_num}', fontsize=16)
    axs[0,0].imshow(control, cmap='gray')
    axs[0,0].set_title('Target static')
    axs[0,0].axis('off')

    axs[0,1].imshow(image, cmap='gray')
    axs[0,1].set_title('unet static')
    axs[0,1].axis('off')

    axs[0,2].imshow(image2, cmap='gray')
    axs[0,2].set_title('pmnetv3 static')
    axs[0,2].axis('off')


    axs[1,0].imshow(control_pl, cmap='gray')
    axs[1,0].set_title('Target postproc')
    axs[1,0].axis('off')

    axs[1,1].imshow(image_pl, cmap='gray')
    axs[1,1].set_title('unet postproc')
    axs[1,1].axis('off')

    axs[1,2].imshow(image2_pl, cmap='gray')
    axs[1,2].set_title('pmnetv3 postproc')
    axs[1,2].axis('off')

    plt.tight_layout()
    plt.show()
    fig.savefig(f'./output/comparisons/wrap{im_num}.png', bbox_inches='tight', dpi=1200)
```

```python
for i in ['002', '004', '018', '052', '053', '054', '055', '056', '057']:
    show_comparisons(i)
```
